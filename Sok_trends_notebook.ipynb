```python
df["year"] = df["year"].astype(int)

# Group by year and category to get publication counts
grouped = df.groupby(["year", "Category"]).size().reset_index(name="count")

colors = {
    "Attack": "#E69F00",                # orange
    "Defense": "#56B4E9",               # sky blue
    "Security and Vulnerability Analysis": "#009E73",  # green
    "Survey": "#F0E442",                # yellow
    "Attack/Defense": "#0072B2"         # blue
}
# Create a list of all years (sorted)
years = sorted(grouped["year"].unique())
# Initialize the figure
fig = go.Figure()

# Add one bar trace per category
for category, color in colors.items():
    category_data = grouped[grouped["Category"] == category]
    fig.add_trace(go.Bar(
        x=category_data["year"],
        y=category_data["count"],
        name=category,
        marker_color=color
    ))

fig.update_layout(
    barmode='group',
    xaxis_title='Year',
    yaxis_title='Publication Count',
    template='simple_white',
    legend_title_text='',
    legend=dict(
        orientation="h",
        yanchor="top",
        y=-0.2,
        xanchor="center",
        x=0.5
    ),
    font=dict(family="Times New Roman", size=14),
    width=1000,
    height=500,
    margin=dict(l=80, r=30, t=30, b=80),
    xaxis=dict(
        tickmode='array',
        tickvals=list(range(2016, 2026)),
        ticktext=[str(year) for year in range(2016, 2026)],
        tickangle=0,
    )
)
fig.show()
```

```python
fig.write_html("publication_bar_chart.html", include_plotlyjs='cdn', full_html=True)
print("publication_bar_chart.html")
```

```python
# Extract ecosystem codes from brackets
import re
def extract_ecosystem_code(text):
    if pd.isna(text):
        return None
    # Find all text in brackets
    matches = re.findall(r'\[([^\]]+)\]', str(text))
    if matches:
        return matches[-1]  # Return the last bracket content (usually the code)
    return None
# Apply extraction
df['Ecosystem_Code'] = df['Ecosystem'].apply(extract_ecosystem_code)
# Show unique codes
print("Unique Ecosystem Codes:")
print(df['Ecosystem_Code'].value_counts())
print("\nTotal ecosystems with codes:", df['Ecosystem_Code'].notna().sum())
```

```python
# Consolidate ecosystem codes into major categories
ecosystem_mapping = {
    # IoT - ALL IoT related (consumer, industrial, etc.)
    'IoT': 'IoT Systems',
    'IIoT': 'IoT Systems',  # Industrial IoT goes to IoT
    'CEIoT': 'IoT Systems',
    'HD': 'IoT Systems',  # Home Devices
    'WN': 'IoT Systems',  # Wireless Networks
    'RI': 'IoT Systems',  # Robotics in IoT context
    'AR': 'IoT Systems',  # Automation/Robotics
    
    # Embedded Systems (Generic/Bare-metal) & System-on-Chip
    'ES': 'Embedded Systems & SoC',
    'Embedded': 'Embedded Systems & SoC',
    'Embedded Systems': 'Embedded Systems & SoC',
    'SoC': 'Embedded Systems & SoC',
    'EC': 'Embedded Systems & SoC',
    'BoF': 'Embedded Systems & SoC',
    
    # Firmware Images & Binary Analysis
    'FI': 'Firmware Images & Binaries',
    'BiF': 'Firmware Images & Binaries',
    'Firmware': 'Firmware Images & Binaries',
    'UEFI': 'Firmware Images & Binaries',  # UEFI is firmware-level
    
    # Critical Infrastructure & Industrial Control Systems
    'CPS': 'Critical Infrastructure & ICS',
    'ICS': 'Critical Infrastructure & ICS',
    'SCADA': 'Critical Infrastructure & ICS',
    'PLC': 'Critical Infrastructure & ICS',
    'SD': 'Critical Infrastructure & ICS',
    'CI': 'Critical Infrastructure & ICS',
    'TI': 'Critical Infrastructure & ICS',
    
    # Manufacturing Systems (Industrial/3D Printing)
    'AM': 'Manufacturing Systems',  # Additive Manufacturing
    '3DP': 'Manufacturing Systems',  # 3D Printing
    
    # Smart Grid & Power Systems (Combined)
    'MG': 'Smart Grid & Power Systems',
    'SC': 'Smart Grid & Power Systems',
    'PS': 'Smart Grid & Power Systems',  # Power Systems
    
    # Automotive & Autonomous Vehicles
    'Automotive': 'Automotive Systems',
    'AS': 'Automotive Systems',
    'AV': 'Automotive Systems',  # Autonomous Vehicles
    
    # Mobile & Desktop Systems
    'MS': 'Mobile Systems',
    'Mobile': 'Mobile Systems',
    'Android': 'Mobile Systems',
    
    # Medical & Health Devices (Combined)
    'MD': 'Medical & Health Devices',
    'HD': 'Medical & Health Devices',  # If it's health-related
    
    # Cloud & Virtualization (Combined)
    'Cloud': 'Cloud & Virtualization',
    'Virt': 'Cloud & Virtualization',
    
    # Aerospace & Space Systems
    'Space': 'Aerospace Systems',
    
    # Research/General → Embedded Systems (most general firmware research applies here)
    'Research': 'Embedded Systems & SoC',
    'General': 'Embedded Systems & SoC',
}
# Apply mapping
df['Ecosystem_Category'] = df['Ecosystem_Code'].map(ecosystem_mapping)

# Show consolidated categories
print("Consolidated Ecosystem Categories:")
category_counts = df['Ecosystem_Category'].value_counts()
print(category_counts)
print("\nTotal categorized:", df['Ecosystem_Category'].notna().sum())
print("Uncategorized:", df['Ecosystem_Category'].isna().sum())
```

```python
# Create pie chart for ecosystem distribution
# Filter out uncategorized entries for cleaner visualization
ecosystem_data = df[df['Ecosystem_Category'].notna()]['Ecosystem_Category'].value_counts().reset_index()
ecosystem_data.columns = ['Category', 'Count']

# Define a color-blind-friendly palette
colors_eco = [
    '#E69F00',  # Orange
    '#56B4E9',  # Sky Blue
    '#009E73',  # Bluish Green
    '#F0E442',  # Yellow
    '#0072B2',  # Blue
    '#D55E00',  # Vermillion
    '#CC79A7',  # Reddish Purple
    '#999999',  # Gray
    '#661100',  # Dark Red
    '#88CCEE',  # Cyan
    '#44AA99'   # Teal
]

# Create pie chart
fig_pie = go.Figure(data=[go.Pie(
    labels=ecosystem_data['Category'],
    values=ecosystem_data['Count'],
    marker=dict(colors=colors_eco),
    textinfo='label+percent',
    textposition='outside',
    textfont=dict(size=18, family="Times New Roman"),
    hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>',
    hole=0  # Set to 0.3 for donut chart if preferred
)])

fig_pie.update_layout(
    title=dict(
        font=dict(family="Times New Roman", size=18, color='black'),
        x=0.5,
        xanchor='center'
    ),
    font=dict(family="Times New Roman", size=14),
    showlegend=True,
    legend=dict(
        orientation="v",
        yanchor="middle",
        y=0.5,
        xanchor="left",
        x=1.05,
        font=dict(size=18)
    ),
    width=1100,
    height=700,
    margin=dict(l=50, r=250, t=100, b=50)
)

fig_pie.show()

# Print summary statistics
print(f"\nTotal papers analyzed: {len(df)}")
print(f"Papers with ecosystem data: {df['Ecosystem_Category'].notna().sum()} ({df['Ecosystem_Category'].notna().sum()/len(df)*100:.1f}%)")
print(f"\nTop 3 Ecosystems:")
for idx, row in ecosystem_data.head(11).iterrows():
    percentage = (row['Count'] / ecosystem_data['Count'].sum()) * 100
    print(f"  {idx+1}. {row['Category']}: {row['Count']} papers ({percentage:.1f}%)")
```
